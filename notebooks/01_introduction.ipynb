#### Creating dataframe from dictionary

```python
dict_data = {
    "Name":["Prajwal","Suvam","Alice","Bob"],
    "Age":[21,24,26,29],
    "Salary":[5000,5000,2953,3941],
    "Department":["IT","HR","Sales","Marketing"]
}

df = pd.DataFrame(dict_data)

df
```

#### Some basic functions

**1. head() function**
- returns the first/top rows
- default is 5 but can be changed

```python
df.head(2)
```

**2. tail() function**
- returns the last/bottom rows
- default is 5 but can be changed

```python
df.tail(2)
```

**3. shape function**
- returns the size of dataframe ie.(row_num,col_num)

```python
df.shape
```

**4.rename() function**
- renames the columns names
- column names must be provided as a key value pair

```python
df1 = df.rename(columns={"Salary":"Monthly_Salary"})
df1
```

- if we want the change to be applied in the original/ the dataframe we are working on we can set inplace=True

```python
df.rename(columns={"Salary":"Monthly_Salary"},inplace = True)
df
```

**4.info() function**
- gives info of the dataframe at basic level

```python
df.info()
```

**5.describe() function**
- provides statistical summary of numerical datas present in dataframe

```python
df.describe()
```
