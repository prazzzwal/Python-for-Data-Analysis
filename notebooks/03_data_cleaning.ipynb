```python
import pandas as pd
```

```python
dirty_data = {
    "student_id": [1, 2, 3, 4, 5, 5, 6, 7, 8],
    "name": ["Ram", "sita", "Gita ", "Hari", "Nita", "Nita", "Aman", "Rita", "Kiran"],
    "age": [21, 22, "twenty", 23, None, None, 24, 22, -1],
    "gender": ["M", "female", "F", "Male", "F", "F", "M", "FEMALE", "M"],
    "math_marks": [78, 85, 69, 90, 74, 74, 88, 81, 76],
    "science_marks": [82, 88, "seventy-two", 91, 79, 79, 87, 84, 80],
    "city": ["Kathmandu", "pokhara ", "Kathmandu", "Lalitpur", "kathmandu", "kathmandu", "Bhaktapur", "Pokhara", "Lalitpur"],
    "enrollment_date": [
        "2023-01-15", "15/01/2023", "2023/01/16",
        None, "2023-01-18", "2023-01-18",
        "2023-01-19", "invalid_date", "2023-01-20"
    ]
}

df = pd.DataFrame(dirty_data)
df
```

#### Removing irrelevant data

```python
df.drop(columns=["student_id"],inplace=True)
df
```

#### Checking for null values

```python
df.isnull()
```

#### Standardizing the text

```python
df['gender'] = df['gender'].str.upper()
df
```

```python
df['gender'] = df['gender'].replace({"MALE":"M",
                                     "FEMALE":"F"})
df['science_marks'] = df['science_marks'].replace({"seventy-two":"72"})

df['age'] = df['age'].replace({"twenty":"20"})
```

```python
df
```

```python
df['name'] = df['name'].str.capitalize()
df['city'] = df['city'].str.capitalize()
df
```

#### Removing Duplicates

```python
df = df.drop_duplicates()
df
```

```python
df['age'] = pd.to_numeric(df['age'],errors='coerce')
```

```python
df
```

```python
df['age'] = df['age'].mask(df['age']<0) # replace with Nan whre the condition is met
df
```

```python
#filling Nan with mean of the age
df['age'] = df['age'].fillna(df['age'].mean())
df
```

```python
df = df.dropna(axis=0)
df
```

```python
df['enrollment_date'] = pd.to_datetime(df['enrollment_date'],errors='coerce')
```

```python
df['enrollment_date'] = df['enrollment_date'].fillna(pd.Timestamp('2023-01-17'))
df
```
