### Creating new columns

```python
df.dropna()
```

```python
df["total_score"] = (df["math_score"] + df["science_score"] + df["english_score"])
df
```

```python
#Insering at specified location

df.insert(
    loc=7,
    column="percentage",
    value=((df["total_score"])/3).round(2) #round(2) dispplays only 2 digits after decimal
)
```

```python
df
```

```python
df["math_science_ratio"] = (df["math_score"]) / (df["science_score"])
df
```

### apply() function

```python
new_df = df.dropna()
new_df
```

```python
new_df["name"] = new_df["name"].apply(lambda x:x.upper())
new_df
```

```python
##Using apply function with user defined function

def grade(score):
    if score>=80:
        return "A"
    elif score>=60:
        return "B"
    else:
        return "C"
    

new_df["GPA"] = new_df["percentage"].apply(grade)

new_df
```

### map() function

```python
#maps the gpa
gpa_mapping = {
    "A":"Excellent",
    "B":"Very Good",
    "C":"Good"
}

# maps male as 1 and female as 0
gender_map = {
    "M":"1",
    "F":"0"
}


new_df["gender"] = new_df['gender'].map(gender_map)

#creating new column remarks
new_df['remarks'] = new_df['GPA'].map(gpa_mapping)

new_df


```


### Binning:

  * `pd.cut`
  * `pd.qcut`



```python
d = {'Student': ['Aryan', 'Prajwal', 'Vishakshi', 'Brijkant', 'Kareena'],
        'Marks': [77, 72, 19, 68, 45]}
data = pd.DataFrame(d)
data
```

```python
bins = [0,50,75,100]
marks_label = ["Low", "Average", "High"]

data['Grade'] = pd.cut(data["Marks"],bins = bins, labels=marks_label,include_lowest=True)
```
