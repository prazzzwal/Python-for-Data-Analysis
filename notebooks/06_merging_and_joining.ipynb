```python
import pandas as pd

users = pd.DataFrame({
    "user_id": [1, 2, 3, 4],
    "name": ["Alice", "Bob", "Charlie", "David"],
    "city": ["Kathmandu", "Pokhara", "Lalitpur", "Bhaktapur"]
})

users
```

```python
orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105],
    "user_id": [1, 2, 2, 4, 5],
    "product": ["Rice", "Wheat", "Sugar", "Oil", "Tea"],
    "amount": [500, 300, 200, 450, 100]
})
orders
```

```python
products = pd.DataFrame({
    "product": ["Rice", "Wheat", "Sugar", "Oil", "Tea", "Coffee"],
    "category": ["Grain", "Grain", "Sweetener", "Oil", "Beverage", "Beverage"],
    "unit_price": [50, 30, 20, 90, 10, 80]
})
products
```

### Inner Join

```python
pd.merge(users,orders,on='user_id',how='inner')
```

```python
pd.merge(orders,products)
```

### Outer/Full Join
- not recommended for ML pipelines

```python
pd.merge(users,orders,how='outer')
```

```python
pd.merge(orders,products,how='outer')
```

### Left Join
- all rows from left and matching from right

```python
pd.merge(orders,products,on='product',how='left')
```

### Right Join
- all rows from right and matching from left

```python
pd.merge(orders,products,on='product',how='right')
```

## Join
- merges dataframe based on index where the default joining is left join

```python
joined = users.join(products)
joined
```

### Joining on keys


```python
new = users.join(products,on='user_id')
new
```

## Concat
- adds the dataframe vertically by default
- the index doesnt gets changed and remains same as in the original dataframe

```python
conc = pd.concat([users,orders],ignore_index=True)
conc
```
