```python
data = {
    "date": pd.date_range(start="2023-01-01", end="2023-03-31", freq="D"),
    "temperature": np.random.randint(15, 35, 90),
    "rainfall": np.random.randint(0, 20, 90),
    "humidity": np.random.randint(40, 90, 90),
    "crop_yield": np.random.randint(20, 60, 90)
}

df = pd.DataFrame(data)
df.head()
```

```python
data = {
    "date": pd.date_range(start="2023-01-01", end="2023-03-31", freq="D"),
    "temperature": np.random.randint(15, 35, 90),
    "rainfall": np.random.randint(0, 20, 90),
    "humidity": np.random.randint(40, 90, 90),
    "crop_yield": np.random.randint(20, 60, 90)
}

df1 = pd.DataFrame(data)
df1
```

```python
df1['date']= pd.to_datetime(df1['date'])
df1.dtypes
```

```python
df1["2023-01-01":"2023-01-10"]
```

```python
df1['3days_temp_avg'] = df1['temperature'].rolling(window=3).mean()
df1
```

### Shifting
- moving data forward or backward along the time axis
- shift(1) -> moves down -> uses past values
- shift(-1) -. modes up -> uses future values

#### Positive Shifting

```python
df1['temp_shift1'] = df1["temperature"].shift(1)
df1
```

#### Negative Shifting

```python
df1['temp_shift_2'] = df1['temperature'].shift(-2)
df1
```
